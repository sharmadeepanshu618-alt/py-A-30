```python
import matplotlib.pyplot as plt
```

```python
student=['a','b','c','d','e','f']
marks=[80,90,45,40,50,60]
```

```python
plt.plot(student,marks)
plt.title('students marks distribution')
plt.xlabel('student names')
plt.ylabel('marks')
plt.show()
```

```python
plt.plot(student,marks,linestyle='--')
plt.title('students marks distribution')
plt.xlabel('student names')
plt.ylabel('marks')
plt.show()
```

```python
plt.plot(student,marks,linestyle='-.')
plt.title('students marks distribution')
plt.xlabel('student names')
plt.ylabel('marks')
plt.show()
```

```python
plt.plot(student,marks,linestyle=':')
plt.title('students marks distribution')
plt.xlabel('student names')
plt.ylabel('marks')
plt.show()
```

```python
plt.plot(student,marks,linestyle='solid')
plt.title('students marks distribution')
plt.xlabel('student names')
plt.ylabel('marks')
plt.show()
```

```python
# dashed  '--'
# dash-dot '-.'
# doted  ':'
# solid  '_'
```

```python
plt.plot(student,marks,linestyle='-.',linewidth=5)
plt.title('students marks distribution')
plt.xlabel('student names')
plt.ylabel('marks')
plt.show()
```

```python
plt.plot(student,marks,linestyle='solid',marker='p')
plt.title('students marks distribution')
plt.xlabel('student names')
plt.ylabel('marks')
plt.show()
```

```python
#marker 'o,.,^,v,s,p'
```

```python
plt.plot(student,marks,linestyle='--',color='r')
plt.title('students marks distribution')
plt.xlabel('student names')
plt.ylabel('marks')
plt.show()
```

```python
plt.plot(student,marks,linestyle='--')
plt.title('students marks distribution')
plt.xlabel('student names')
plt.ylabel('marks')
plt.grid()
plt.show()
```
