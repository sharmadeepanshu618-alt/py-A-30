```python
try:
    a=int(input("enter numerator:"))
    b=int(input("enter denominator:"))
    print("the division of a/b is=",a/b)
except:
    print("you cannot enter denominator as 0")
    b=int(input("enter denominator:"))
    if b!=0:
        print("the division of a/b is=",a/b)
    else:
        print("the answer goes to infinity,try again")
        

         
```

```python
c=600
try:
    print(c)
except:
    print("c in not yet initialised,plese enter c:")
    c=input("please enter c's value'")
    print("this is the value of c:",c)
else:
    print ("this is else block")
    #the else block runs when the try block throws no error while running.
```

```python
d=10
try:
    print(d)
except:
    print("d in not yet initialised,plese enter d:")
    d=input("please enter d's value'")
    print("this is the value of d:",d)
else:
    print ("this is else block")
finally:
    print("this is finally block")
    #the finally block always runs no matter if the try block throws an error or not
```
