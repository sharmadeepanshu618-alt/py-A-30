```python
import matplotlib.pyplot as plt
```

```python
"""
subplot()
the subplot fun takes 3 arguments that describe the layout of the figure.
the layout is organised in rows and col. which are represented by the first and second argument
the third argu. represents the index of the current plot.
example:
        1.plt.subplot(1,2,1)
        here the figure has 1 row 2 colu. and this plot is the first plot as the last argument is 1
        2.plt.subplot(1,2,2)
        here the figure has 1 row 2 column and this plot is the
         second plot as the last argument is 2.
"""         
```

```python
#plot1
student=['a','b','c','d','e','f']
height=[163,145,167,180,178,177]
plt.title('height')
plt.subplot(1,2,1)
plt.plot(student,height)

#plot2
student=['a','b','c','d','e','f']
weight=[67,89,45,67,90,88]
plt.title('weight')
plt.subplot(1,2,2)
plt.plot(student,weight)

plt.show()
```

```python
student=['a','b','c','d','e','f']
height=[163,145,167,180,178,177]

plt.subplot(2,2,1)
plt.plot(student,height)
plt.title('height')

#plot2
student=['a','b','c','d','e','f']
weight=[67,89,45,67,90,88]

plt.subplot(2,2,2)
plt.plot(student,weight)
plt.title('weight')

student=['a','b','c','d','e','f']
marks=[77,89,90,56,66,88]

plt.subplot(2,2,3)
plt.plot(student,marks)
plt.title('marks')


student=['a','b','c','d','e','f']
age=[22,24,19,22,20,21]

plt.subplot(2,2,4)
plt.plot(student,age)
plt.title('age')
plt.suptitle('student statistics')
plt.subplots_adjust (wspace=0.5,hspace=0.5)

plt.show()

```
