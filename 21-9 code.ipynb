```python
def changecase(func):#decorator
  def myinner():
      return func().upper()
  return myinner
  
  

@changecase
def myfunction():  # main function
  return "Hello Sally"

print(myfunction())
```

```python
def changecase(func):
  def myinner():
    return func().upper()
  return myinner

@changecase
def myfunction():
  return "Hello Sally"

@changecase
def otherfunction():
  return "I am speed!"

print(myfunction())
print(otherfunction())
```
