```python
#pie chart
# a pie chart show how a total is divided among diffrent categories

values=[30,40,50,20,15]
students=['a','b','c','d','e']
color=['orange','blue','cyan','brown','red']
plt.pie(values,labels=students,autopct="%1.1f%%",shadow= True,colors=color)
plt.show()
```

```python
# plot pie charts for all the following data in one place using subplot
# Set 1 Labels & Numerical Values
labels_smartphone = ["Apple", "Samsung", "Xiaomi", "Oppo", "Others"]
sizes_smartphone = [21.5, 19.8, 13.2, 8.5, 37.0]

# Set 2 Labels & Numerical Values
labels_budget = ["Housing", "Food & Groceries", "Transportation", "Savings", "Entertainment"]
sizes_budget = [35, 20, 15, 20, 10]

# Set 3 Labels & Numerical Values
labels_atmosphere = ["Nitrogen", "Oxygen", "Argon", "Carbon Dioxide", "Neon & Others"]
sizes_atmosphere = [78.08, 20.95, 0.93, 0.04, 0.00] 

# Set 4 Labels & Numerical Values
labels_gaming = ["PC", "PlayStation", "Xbox", "Nintendo Switch", "Mobile"]
sizes_gaming = [24, 22, 18, 16, 20]


plt.subplot(2, 2, 1)
plt.pie(
    sizes_smartphone,
    labels=labels_smartphone,
    autopct="%1.1f%%",
    shadow=True
)
plt.title("Smartphone Market Share")


# Chart 2
plt.subplot(2, 2, 2)
plt.pie(
    sizes_budget,
    labels=labels_budget,
    autopct="%1.1f%%",
    shadow=True
)
plt.title("Monthly Budget")


# Chart 3
plt.subplot(2, 2, 3)
plt.pie(
    sizes_atmosphere,
    labels=labels_atmosphere,
    autopct="%1.2f%%",
    shadow=True
)
plt.title("Atmosphere Composition")


# Chart 4
plt.subplot(2, 2, 4)
plt.pie(
    sizes_gaming,
    labels=labels_gaming,
    autopct="%1.1f%%",
    shadow=True
)
plt.title("Gaming Platform")


plt.tight_layout()
plt.show()
```
